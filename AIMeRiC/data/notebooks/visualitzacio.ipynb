# Visualització de dades

En aquest quadern es duen a terme diverses operacions per visualitzar les dades obtingudes de l'escansió automàtica amb _AiMeRiC_. 

```python
import pandas as pd
from AIMeRiC.corpus.modules.getUSB import getUSB
import AIMeRiC.data.modules.visualitza as vis

usb = getUSB()
corpus = 'actual'
metadades = f'{usb}/corpora/{corpus}/etiquetat/xlsx/metadades_autors.xlsx'
df_metadades = pd.read_excel(metadades)
corpus_path = f'{usb}/corpora/{corpus}/etiquetat/{corpus}.csv'
df_corpus = pd.read_csv(corpus_path, sep=';', encoding='utf-8')
```

## Autors i mida del corpus

```python
vis.desctiptives(df_corpus, corpus)
```

## Origen dels textos del corpus

```python
vis.font(df_metadades)
```

##  Origen geogràfic dels autors

```python
import geopandas as gpd

path = f'{usb}/corpora/{corpus}/etiquetat/svg/mapa.geojson'
gdf = gpd.read_file(path)
vis.mapa_estatic(gdf)
```

## Autors en l'eix cronològic

```python
vis.anys(df_metadades)
```

## Autors per gènere

```python
vis.generes(df_metadades)
```
