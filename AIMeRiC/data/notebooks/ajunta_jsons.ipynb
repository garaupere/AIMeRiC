# Ajunta jsons
Un breu codi per ajuntar tots els json resultants de l'etiquetatge en un únic fitxer.

```python
from datetime import datetime
import json

import os

path = 'D:/corpora/actual/etiquetat/json/'

global_json = {
        'name': 'Corpus del decasíl·lab català contemporani (CDCC)',
        'compiler': 'Pere Garau',
        'description': 'Corpus del decasíl·lab català contemporani (CDCC) és un corpus de decasíl·labs catalans etiquetats mètricament.',
        'tagger': 'AiMeRiC',
        'date': datetime.now().strftime('%Y-%m-%d %H:%M:%S'),
        'authors': {},
        }
for file in os.listdir(path):
    if file.endswith('.json'):
        with open(os.path.join(path, file), 'r', encoding='utf-8') as f:
            data = json.load(f)
            print(data['name'])
            global_json['authors'].update({data['name']: data})

with open('D:/corpora/actual/etiquetat/actual.json', 'w', encoding='utf-8') as f:
    json.dump(global_json, f, ensure_ascii=False, indent=4)
```
